```python

from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_DIR = Path(
    r"C:\Users\Dell5420\OneDrive\Desktop\SaaS_GTM_Customer_Analytics"
)

RAW_DIR = PROJECT_DIR / "data" / "raw"

csv_files = list(RAW_DIR.glob("*.csv"))

print("CSV files found:", len(csv_files))

for file in csv_files:
    print(file.name)
```

```python

# Load the main marketing dataset
file_path = RAW_DIR / "marketing_data.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
```

```python

# View the first five rows
display(df.head())

# Check column names
print("\nColumn names:")
print(df.columns.tolist())

# Inspect data types and non-null values
print("\nDataset information:")
df.info()
```

```python

# Count missing values in every column
missing_values = df.isnull().sum()

print("Missing values by column:")
print(missing_values)

print("\nTotal missing cells:", df.isnull().sum().sum())
```

```python

# Remove leading and trailing spaces from all column names
df.columns = df.columns.str.strip()

# Confirm the Income column now exists
print("Income" in df.columns)
```

```python

df[df["Income"].isnull()]
```

```python

# Check for completely duplicated rows
duplicate_rows = df.duplicated().sum()

print("Completely duplicated rows:", duplicate_rows)

```

```python

# Check for duplicate customer IDs
duplicate_ids = df["ID"].duplicated().sum()

print("Duplicate customer IDs:", duplicate_ids)
```

```python

# Check the data type of every column
print(df.dtypes)
```

```python

# Check the current data type
print("Current data type:", df["Dt_Customer"].dtype)

# Inspect a few date values
print("\nSample values:")
print(df["Dt_Customer"].head())
```

```python

# Convert Dt_Customer from text to datetime
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    errors="coerce"
)

# Validate the conversion
print("New data type:", df["Dt_Customer"].dtype)

print(
    "Missing dates after conversion:",
    df["Dt_Customer"].isnull().sum()
)

display(df[["ID", "Dt_Customer"]].head())
```

```python

# Summary statistics for income
print(df["Income"].describe())

# Display customers with missing income
display(
    df.loc[
        df["Income"].isnull(),
        ["ID", "Education", "Marital_Status",
         "Kidhome", "Teenhome", "Dt_Customer"]
    ]
)
```

```python

# Display the 10 customers with the highest incomes
df.nlargest(10, "Income")[
    ["ID", "Income", "Education", "Marital_Status"]
]
```

```python

# Calculate the first and third quartiles
Q1 = df["Income"].quantile(0.25)
Q3 = df["Income"].quantile(0.75)

# Calculate the interquartile range
IQR = Q3 - Q1

# Calculate the upper outlier threshold
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Upper outlier threshold:", upper_limit)

# Count income values above the threshold
outliers = df[df["Income"] > upper_limit]

print("Potential high-income outliers:", len(outliers))
```

```python

# Display the eight potential income outliers
outliers[
    ["ID", "Income", "Education", "Marital_Status",
     "Year_Birth", "Kidhome", "Teenhome"]
].sort_values("Income", ascending=False)
```

```python

# Calculate the original mean income
original_mean = df["Income"].mean()

# Identify the maximum income
max_income = df["Income"].max()

# Calculate a comparison mean excluding only the maximum
comparison_mean = df.loc[
    df["Income"] < max_income, "Income"
].mean()

print("Original mean income:", round(original_mean, 2))
print("Maximum income:", max_income)
print("Comparison mean without maximum:", round(comparison_mean, 2))
print(
    "Difference in means:",
    round(original_mean - comparison_mean, 2)
)
```

```python

# Compare mean and median income
income_mean = df["Income"].mean()
income_median = df["Income"].median()

print("Mean income:", round(income_mean, 2))
print("Median income:", round(income_median, 2))

print("\nMissing income before treatment:",
      df["Income"].isnull().sum())
```

```python

# Create a working copy; preserve the original df
df_test = df.copy()

# Record which income values were originally missing
df_test["Income_Was_Missing"] = df_test["Income"].isnull()

# Calculate the median using observed income values
income_median = df_test["Income"].median()

# Fill missing income in the test copy only
df_test["Income"] = df_test["Income"].fillna(income_median)

# Validate the result
print("Original missing incomes:", df["Income"].isnull().sum())
print("Missing incomes after imputation:",
      df_test["Income"].isnull().sum())
print("Median used:", income_median)
print("Original row count:", len(df))
print("Test copy row count:", len(df_test))
```

```python

# Count customers whose income was originally missing
print(
    "Originally missing incomes flagged:",
    df_test["Income_Was_Missing"].sum()
)

# Display a few of those records
display(
    df_test.loc[
        df_test["Income_Was_Missing"],
        ["ID", "Income", "Income_Was_Missing"]
    ].head()
)
```

```python

# Inspect all numerical columns
display(df.describe().T)
```

```python

# Inspect the 10 earliest birth years
df.nsmallest(10, "Year_Birth")[
    ["ID", "Year_Birth", "Education", "Marital_Status", "Income"]
]
```

```python

# Count customers born before 1940
early_birth_years = df[df["Year_Birth"] < 1940]

print("Customers born before 1940:", len(early_birth_years))

display(
    early_birth_years[
        ["ID", "Year_Birth", "Education", "Marital_Status", "Income"]
    ].sort_values("Year_Birth")
)
```

```python

# Create a working copy for further data-quality checks
df_clean = df.copy()

# Flag potentially suspicious birth years
df_clean["Birth_Year_Flag"] = df_clean["Year_Birth"] < 1940

print("Records flagged for birth-year review:",
      df_clean["Birth_Year_Flag"].sum())

print("\nFlagged records:")
display(
    df_clean.loc[
        df_clean["Birth_Year_Flag"],
        ["ID", "Year_Birth", "Birth_Year_Flag"]
    ]
)
```

```python

# Inspect unique categories in important text columns

categorical_columns = ["Education", "Marital_Status", "Country"]

for col in categorical_columns:
    print(f"\n{col} — unique values:")
    print(sorted(df_clean[col].dropna().astype(str).unique()))

    print("Number of unique categories:", df_clean[col].nunique())
```

```python

# Count customers in each marital-status category
print(df_clean["Marital_Status"].value_counts(dropna=False))

# Inspect records with unusual labels
unusual_statuses = ["Absurd", "YOLO"]

display(
    df_clean.loc[
        df_clean["Marital_Status"].isin(unusual_statuses),
        ["ID", "Education", "Marital_Status", "Income", "Year_Birth"]
    ]
)
```

```python

# Define categories that need review
suspicious_statuses = ["Absurd", "YOLO"]

# Flag records with suspicious marital-status labels
df_clean["Marital_Status_Flag"] = (
    df_clean["Marital_Status"].isin(suspicious_statuses)
)

print(
    "Records flagged for marital-status review:",
    df_clean["Marital_Status_Flag"].sum()
)

display(
    df_clean.loc[
        df_clean["Marital_Status_Flag"],
        ["ID", "Marital_Status", "Marital_Status_Flag"]
    ]
)
```

```python

# Columns that should not contain negative values
non_negative_columns = [
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds",
    "NumDealsPurchases",
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases",
    "NumWebVisitsMonth",
    "Kidhome",
    "Teenhome"
]

# Count negative values in each column
negative_counts = (df_clean[non_negative_columns] < 0).sum()

print("Negative values per column:")
print(negative_counts)

print("\nTotal negative cells:", negative_counts.sum())
```

```python

# Campaign and response indicator columns
binary_columns = [
    "AcceptedCmp1",
    "AcceptedCmp2",
    "AcceptedCmp3",
    "AcceptedCmp4",
    "AcceptedCmp5",
    "Response",
    "Complain"
]

# Inspect the distinct values in each column
for col in binary_columns:
    print(f"{col}: {sorted(df_clean[col].dropna().unique())}")

# Check whether any values fall outside 0 and 1
invalid_binary_counts = (
    ~df_clean[binary_columns].isin([0, 1])
).sum()

print("\nInvalid values per column:")
print(invalid_binary_counts)

print("\nTotal invalid cells:", invalid_binary_counts.sum())
```

```python

# Display columns and their data types
print(df_clean.dtypes)
```

```python

# Summarise our data-quality findings

print("DATA QUALITY SUMMARY")
print("-" * 35)

print("Total records:", len(df_clean))
print("Total columns:", len(df_clean.columns))

print("\nMissing values in original dataset:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nDuplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["ID"].duplicated().sum())

print("\nBirth years flagged for review:",
      df_clean["Birth_Year_Flag"].sum())

print("Marital statuses flagged for review:",
      df_clean["Marital_Status_Flag"].sum())

print("\nNegative values in checked columns:",
      (df_clean[non_negative_columns] < 0).sum().sum())

print("Invalid campaign indicators:",
      (~df_clean[binary_columns].isin([0, 1])).sum().sum())
```

```python

# Define the processed-data folder
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"

# Create the folder if it doesn't already exist
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Save the working dataset as a CSV
output_path = PROCESSED_DIR / "marketing_data_cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Dataset saved successfully!")
print("File location:", output_path)
print("Rows and columns:", df_clean.shape)
```

```python

# Reload the saved file to verify it
df_saved = pd.read_csv(output_path)

print("Saved file exists:", output_path.exists())
print("Saved dataset shape:", df_saved.shape)
print("Missing income values:", df_saved["Income"].isnull().sum())
print("Birth-year flags:", df_saved["Birth_Year_Flag"].sum())
print("Marital-status flags:", df_saved["Marital_Status_Flag"].sum())
```
